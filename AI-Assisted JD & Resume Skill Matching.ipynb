```python
# AI-Assisted JD & Resume Skill Matching System


#Install Required Libraries
!pip install spacy scikit-learn pandas
!python -m spacy download en_core_web_sm

#Import Libraries
import spacy
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

#Load NLP Model
nlp = spacy.load("en_core_web_sm")

#Sample Job Description & Resume
job_description = """
We are looking for a Data Analyst with strong skills in Python, SQL, Excel,
Power BI, data visualization, and statistical analysis.
"""

resume_text = """
Experienced analyst skilled in Python, Excel, SQL, Pandas, and Power BI.
Worked on dashboards, reporting, and data analysis projects.
"""
#Skill Extraction Function
skills_list = [
    "python", "sql", "excel", "power bi", "tableau",
    "machine learning", "statistics", "data analysis",
    "pandas", "numpy"
]

def extract_skills(text):
    text = text.lower()
    extracted = [skill for skill in skills_list if skill in text]
    return extracted


#Extract Skills from JD & Resume
jd_skills = extract_skills(job_description)
resume_skills = extract_skills(resume_text)

print("JD Skills:", jd_skills)
print("Resume Skills:", resume_skills)

# Define matched_skills before it's used
matched_skills = set(jd_skills).intersection(set(resume_skills))

match_percentage = (len(matched_skills) / len(jd_skills)) * 100

print("Matched Skills:", matched_skills)
print(f"Skill Match Percentage: {match_percentage:.2f}%")

documents = [job_description, resume_text]

#Semantic Similarity Using AI
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)

similarity_score = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])[0][0]

print(f"JD–Resume Similarity Score: {similarity_score:.2f}")

#Final AI Match Decision

if similarity_score > 0.6 and match_percentage > 60:
    print("✅ Strong Match – Resume is suitable for the role")
elif similarity_score > 0.4:
    print("⚠ Partial Match – Resume needs skill improvement")
else:
    print("❌ Weak Match – Resume not suitable")

#Future enhancement
 #1.we can upload multiple resume and check for match in JD
```
